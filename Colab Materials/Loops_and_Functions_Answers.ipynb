### If Statement (Conditional)

The `if` statement:

-   runs code based on whether a condition is met
-   if condition is met, code will be run once

The expression you use for your condition must evaluate to `TRUE` or `FALSE`. 
Examples of these types of expressions include:
- boolean/logical variables
- relational expressions e.g. "mynumber > 5"
- logical expressions e.g. "isHelpful && isAvailable"


```r

# Use a boolean variable
isWolf <- TRUE

if (isWolf) {
  print("This is a wolf.")
}
```

### For Loop

A `for` loop:

-   runs a specific number of times

-   will run at least once

-   is useful for traversing data structures like vectors, matrices, and dataframes.

```r
# FOR loop over a range
for (i in 1:5) {
  print(i) # spice it up and divide i by 2
}
```

```r
# FOR loop over elements in a vector (can be any dataype)
vector2 = c(2, 4, 6, 8, 10)
for (y in vector2) {
  print(y**y)
}
```

We can create nested `for` loops.

```r
# A nested for loop
for (i in 1:4) {
    print(paste("i=",i))
    print('* * * * * * *')
    for (j in 1:3) {
        print(paste("j=",j))
    }
}
```

We can use nested loops to iterate over a two-dimensional matrix or data frame.

```r
# Create some data
data = rnorm(30, 64, 6)

# Use `data` to create a 2-D matrix from the data
wtWolves = matrix(data, nrow=5, ncol=6, byrow = TRUE)
print("Wolves matrix: ")
print(wtWolves)
```

```r
# Use nested for loop to print out every value in the matrix
print("Elements in Wolves matrix")
for (row in 1:5) {   # Why isn't this working?
  for (col in 1:6) {
    print(wtWolves[row, col])
  }
}
```

```r
# We can use functions nrow() and ncol() to extract number of rows and cols
# from the data structure
for (row in 1:nrow(wtWolves)) {
  for (col in 1:ncol(wtWolves)) {
    print(wtWolves[row,col])
  }
}
```

```r
# How would you use this same kind of structure to access data in a 3-D array?

# put data in a 3d matrix/array
wtWolves3D <- array(data, dim=c(5,3,2))

# print out dimensions of array
array.dim <- dim(wtWolves3D)

# notice array.dim is a vector with 3 elements
print(array.dim) 
print(wtWolves3D)
```

```r
# Create nested for loop to iterate over, rows, columns and dimensions
# Instead of using nrow() or ncol() to define the end range
# we use the element of `array.dim` that corresponds to the number of
# rows, columns and dimensions in our 3-D array.

for (row in 1:array.dim[1]) {
  for (col in 1:array.dim[2]) {
    for (dim in 1:array.dim[3]) {
      # show values for row, col, dim
      print(paste('row =',row, 'col =', col, 'dim =', dim))
      # print value at row, col, dim
      print(wtWolves3D[row, col, dim])
    }
  }
}
```

### While Loops

A while loop:

-   executes code by testing a condition
-   code may execute once, several times or not at all
-   can be used to run code in response to a changing condition
-   the condition can be a relational expression or logical expression
-   the condition can be a logical variable, however, we'll see that this can be inefficient

NOTES:
- the variables associated with the condition being tested must be set before the while loop runs
- the condition must be updated within the while loop

```r
# Using a relational expression
numWolves <- 5

while (numWolves != 0) {
  print(paste('Number of wolves: ', numWolves))

  # update condition variable
  numWolves = numWolves - 1 # A VERY IMPORTANT STEP
}
```

```r
# Using a logical expression
numWolves <- 15
numRabbits <- 2

while (numWolves > 10 && numRabbits <= 20) {
    print(paste("# of Wolves: ", numWolves))
    print(paste("# of Rabbits: ", numRabbits))

    # update condition variables
    numWolves <- numWolves - 1
    numRabbits <- numRabbits*2
}
```

The code below shows a broken `while` loop. Notice that after you click on the play button you see a stop button in this place. Click on this to stop code from running.

```r
# Broken While loop

numWolves <- 5

while (numWolves != 0) {
  print(paste('Number of wolves: ', numWolves))
}

# numWolves never gets updated!!
```

Good code is:

-   code that works and

-   that can be understood by others.

Efficiency comes with practice and in exploring different ways to write the same code.

Let's look at some code and see how we can make it more efficient.

```r
# Inefficient version of while loop

# Initialize variables
happyWolves <- TRUE
numWolves <- 16

while (happyWolves == TRUE) {
  # A wolf leaves
  numWolves = numWolves - 1

  # the number of wolves drops below five, they get unhappy
  if (numWolves <= 5) {
    happyWolves = FALSE
  }
  print(paste('Number of wolves: ', numWolves))
}

# Walk through the steps in your head. Is there anything that seems 
# duplicative? Focus on the "if" statement used above to change the 
# value of "eatsHamburger".
```

```r
# More efficient while loop

# whether a wolf is happy or not is dependent on the number of wolves
# therefore we don't need happyWolves for the loop
numWolves <- 16

# we can directly check number of wolves, but we need to change the 
# relational operator (why?)
while (numWolves > 5) {
  # a wolf leaves
  numWolves = numWolves - 1
  print(paste('Number of wolves: ', numWolves))
}

# check outputs for the inefficient version and this one.
# Are the outputs the same?
```

Inefficient while loops are likely to occur if you use a logical variable for the while loop test condition. The logical variable will have to be updated using a logical expression (using relational or logical operators). It is usually more efficient to use the logical expression directly than use a logical variable that is updated using the logical expression. However, you may decide to use the inefficient code to maintain readability/interpretability, i.e. make clear to other users what your code is doing.

Some important considerations with `while` loops:

-   make sure you initialize condition values BEFORE the while loop starts

-   make sure values in the test condition are getting updated appropriately within the loop

## Functions

We will look at three types of functions:

-   Built-in

    -   are available when R loads without further info

-   External package functions

    -   need to be initially installed

    -   when you want to use them you must load them using the `library()` command

-   User-defined functions

We'll talk about each of these and explore some examples. The following vocabulary will be helpful:

> **define**: before a function can be used it must be defined. Built-in and external package functions are defined. When you create your own functions (user-defined) you will create a definition.
>
> **call**: when you use a function in your program you "call" it. You can't call a function before it is defined.

A function has three main components:

-   name (so you can refer to it because "hey, you!" won't get its attention)

-   parameters

    -   pieces of information the function needs to do it's job.

    -   parameters can be required or optional

        -   the function won't run without required parameters

        -   optional parameters will have a default value which is used if no value is passed for it

        -   the term parameters refers to the variables the function needs to do its job.

        -   when we provide values for those parameters, we call them arguments.

-   return value

    -   this is a value that the function returns when it's done
    -   frequently we will store this value in a variable using the assignment operator
    -   the return value may be
        -   a value that was calculated
        -   data that has been manipulated in some way, or
        -   a status that indicates whether the function encountered any errors

### User-defined Functions

Functions should:

-   perform a discrete task or set of tasks

-   be written to be reusable

Whether you are just starting out or a seasoned programmer, spending time with paper and pencil before programming is a good idea. Consider:

-   what will the function do?

-   what information (parameters) does it need to do it

-   what parameters will be required vs. optional

    -   it's ok to make everything required

-   what will the function return?



Let's make a function that will calculate the mean of a vector of numbers.

Let's recall how we calculate the mean of a set of numbers. If we have $N$ numbers, we find the mean by adding up all $n$ numbers and then dividing by $N$. 

Let's consider the following before we start writing code:

1.  What are the steps we need to take to calculate a mean? Think in terms of using a calculator or pen and paper.
2.  What information do we need to pass to the function to calculate a mean?
3.  What information do we expect to return once we've calculated the mean?
4.  What data types or structures would be appropriate?
    -   for parameters
    -   for return values
5.  Looking at our steps would we need one of our flow control statements:
    -   In the steps we articulated above, do we need to run code conditionally?
    -   Are there any steps that require us to do anything a preset \# of times?
    -   Do we have steps that need to be repeated but we aren't sure how many?

Base R has a function to calculate the mean, `mean()`. We can use it to confirm our function is calculating the mean correctly.

Now, we probably have enough information to start writing our function. For readability, let's use comments to document some information about our function. Also, let's call our function `avg` since the mean function already exists.

```r
# Function: avg(x, N)
# Required parameters:
#   * x: a vector of numbers
#   * N: the number of elements in the vector
# Return value: avg returns the mean of the numbers in the vector.

avg <- function(x, N){
  sumX <-  0
  for (i in 1:N) {
    sumX <- sumX + x[i]
  }
  avgX <- sumX/N
  return(avgX)
}
```

```r
# Let's Test our Function
# Define some vectors
vector.1 <- runif(50, min=0, max=100) # pulls 50 values from a uniform distribution
vector.2 <- rnorm(50, 0, 1) # pulls 50 numbers from a normal distributions

print(avg(vector.1, 50))
print(mean(vector.1))
print(avg(vector.2, 50))
print(mean(vector.2))
```

There is almost never only one way to write a function. Looking at our solution, we can ask a couple of questions:
1. Do we need to pass $N$ to our function, or is there a base R function that tells us how many elements are in a vector?
2. Do we need to use a `for` loop to calculate the sum of the numbers in our vector or is there a functon in base R that will give me the sum of the elements in a numerical vector?
   
If we don't know the answer, this is where doing a search on the internet or even using an AI assistant could be useful. 

The answers to our questions are:
1. There is a function that returns the length of a vector. If we use it in our code, we don't need to pass the length as a parameter.
2. There is a function that will return the sum of all the elements of a numeric vector (will it accept a vector of strings?). We can replace our `for` loop with a single line of code.

```r
# Second iteration of avg() function
avg2 <- function(x){ 
  N <- length(x)
  sumX <-  sum(x)
  avgX <- sumX/N
  return(avgX)
}

```

```r
# Test avg2 against avg and mean functions
print(avg2(vector.1))
print(avg(vector.1, 50))
print(mean(vector.1))

print(avg2(vector.2))
print(avg(vector.2, 50))
print(mean(vector.2))
```

### Practice Writing Your Own Function

Your turn!! Try writing a function to calculate standard deviation of a vector of numbers. Recall that the steps for calculating the **s**tandard **d**eviation are:

1. Calculate the mean of the vector. (Is there a function that can do this?)
2. For each element in the vector subtract the mean and square the result 
3. Generate a sum of all the values calculated in step 2
4. Divide the sum by the number of elements in the original vector minus one.
5. Take the square root of the value calculated in step 4

Name your function "stndev" and test it using the R `sd()` function. Before writing code, walk through the steps listed above. You can use the `sqrt()` function to calculate the square root.

#### Example Solution:

1.  What are the steps we need to take to calculate a standard deviation?

    -   We need to:
        -   Calculate the mean of the numbers.
        -   For each number, we need to subtract the mean from it and then square the result
        -   We need to add this value to a running sum
        -   Once we have calculated the sum, we need to take the square root of the sum
        -   We then need to divide the sum by the number of values - 1
        -   Finally, we need to take the square root of the value calculated in the previous step

2.  What information do we need to pass to the function to calculate a standard deviation?

    -   a vector of numerical values
    -   the number of values passed

3.  What information do we expect to return once we've calculated the mean?

    -   the standard deviation

4.  What data types or structures would be appropriate?

    -   for parameters: a vector of numerical values
    -   for return values: a number representing the standard deviation

5.  Looking at our steps would we need one of our flow control statements:

    -   In the steps we articulated above, do we need to run code conditionally? No.
    -   Are there any steps that require us to do anything a preset \# of times?
        -   we can use our avg() function to calculate the mean
        -   we need to subtract the mean from each of our numbers, square the result and add the result to a variable. This sounds like a `for` loop
        -   we can divide the result by N-1 with a single line of code
        -   we can take the square root of that value with a single line of code
    -   Do we have steps that need to be repeated but we aren't sure how many? No.


```r
# Write an implementation for `stndev()` function

standev <- function(x, N){
  # calculate mean
  my.avg <- avg(x, N)
  # initialize our running sum
  my.sum <- 0
  for (num in x) {
    # subtract mean from our number and square it
    my.value <- (num-my.avg)**2
    # add this value to our running sum
    my.sum <- my.sum + my.value
  }
  my.variance <- my.sum / (N-1)
  my.standev <- sqrt(my.variance)
  return(my.standev)
}
```

```r
# test your function
# Define some vectors
vector.1 <- runif(50, min=0, max=100) # pulls 50 values from a uniform distribution
vector.2 <- rnorm(50, 0, 1) # pulls 50 numbers from a normal distributions

# compare output from your function to the R function `sd()`
print(standev(vector.1, 50))
print(sd(vector.1))
print(standev(vector.2, 50))
print(sd(vector.2))
```
